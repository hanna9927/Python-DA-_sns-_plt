```python
# 1) Loading the Taxis Dataset
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
df.shape

```

```python
df.head(10)
```

```python
df.tail(10)
```

```python
df.info()
```

```python
# 2) Handling Missing Values
```

```python
df.isnull()  
```

```python
df.isnull().sum()  
```

```python
df[['payment', 'pickup_zone', 'dropoff_zone',
    'pickup_borough', 'dropoff_borough']].mode()
```

```python
df['payment'] = df['payment'].fillna(df['payment'].mode()[0])  
df.dropna(subset=['pickup_zone', 'dropoff_zone',
                  'pickup_borough', 'dropoff_borough'], inplace=True)    
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
# 3) Visualizations using Matplotlib/Pandas Plot:
```

```python
print(df['pickup'].dtype)
```

```python
df = df.sort_values('pickup')
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df['pickup'], df['fare'])
plt.xlabel('Pickup Time')
plt.ylabel('Fare')
plt.title('Fare Over Time')
plt.xticks(rotation=45)  
plt.grid(True)

plt.show()
```

```python
borough_fare = df.groupby('pickup_borough')['fare'].sum()
borough_fare
```

```python
borough_fare.plot(kind='bar', figsize=(6, 3))

plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.title('Total Fare by Pickup Borough')

plt.xticks(rotation=45)
plt.show()
```

```python
payment_count = df['payment'].value_counts()
payment_count
```

```python
payment_count.plot(kind='pie', autopct='%1.1f%%', figsize=(6, 3))

plt.title('Distribution of Trips by Payment Method')
plt.ylabel('')

plt.show()

```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["distance"],bins= 15,color ="purple",edgecolor='red')

plt.xlabel('Distance')
plt.ylabel('Frequency')
plt.title('Distribution of Taxi Trip Distance')
plt.show()
```

```python
plt.figure(figsize=(6, 3))

df.boxplot(column='tip', by='pickup_borough',color = "green")
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')
plt.title('Distribution of Tips by Pickup Borough')
plt.suptitle('')

plt.show()
```

```python
plt.figure(figsize=(6, 3))
sns.countplot(data=df, x='pickup_borough', color='yellow')
plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')
plt.title('Number of Trips by Pickup Borough')
plt.xticks(rotation=45)
plt.show()

```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df,x='distance',y='fare',hue='pickup_borough')
plt.xlabel('Distance')
plt.ylabel('Fare')
plt.title('Relationship Between Distance and Fare')

plt.show()
```

```python
numeric_df = df[['distance', 'fare', 'tip', 'tolls', 'total']]
correlation = numeric_df.corr()
correlation
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Taxi Dataset')

plt.show()

```

```python
top_zones = df['pickup_zone'].value_counts().head(5).index
top_zones
```

```python
pair_df = df[df['pickup_zone'].isin(top_zones)]
pair_df = pair_df[['distance', 'fare', 'tip', 'total', 'pickup_zone']]
sns.pairplot(pair_df,hue='pickup_zone')

```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='payment', y='fare',hue='payment',legend=False,palette='Set2')
plt.xlabel('Payment Method')
plt.ylabel('Fare')
plt.title('Fare Distribution by Payment Method')

plt.show()

```
